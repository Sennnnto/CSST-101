**Creating a Simple Rule-based System**

```r
classify_risk <- function(attendance,grade)
if (attendance <= 75 && grade <= 75){
  print("High risk")
} else{
  "Low Risk"
}

classify_risk(85,91)
```

**Adding More Rules**

```r
student_risk <- function(attendance,grade)
if (attendance < 75 && grade < 75){
  print("High risk")
}else if (attendance < 85 || grade < 80){
  print("Moderate risk")
}else if (attendance >= 85 && grade >= 80){
  "Low Risk"
}

student_risk(69,2)
```

**Apply The Rule to Multiple Students**

```r
students$risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

students
```
